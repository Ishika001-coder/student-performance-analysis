```python
import seaborn as sns
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot = True, cmap = "coolwarm", fmt = ".2f")
plt.title("Correlation Heatmap of student performance factors")
plt.show()
```

#key findings

-The data set contains 14003 student records and 16 attributes.
-There are no missing values in the dataset.
-The average exam scores is approximately 70.35.
-The average attendance is approximately 80.19%.
-The average study Hours are approximately 19.99 hours.
-The highest exam score is 100.
-The lowest exam score is 40.
-Motivation level 0, 1 and 2 have average exam scores of approximately 70.28, 70.48 and 70.12 respectively.
-Stress level 0, 1 and 2 have average exam scores of approximately 71.88, 69.75 and 70.08 respectively.

#Conclusion
This project analyzed the student performance and learning behaviour dataset containing 14003 student records and 16 attributes. The analysis found that the average exam score was 70.35, while the average attendance was 80.19% and average study hours 19.99 the dataset contained no missing values.
Different visualisation were used to examine student performance in relation to study hours, assignment completion, attendance, motivation, stress level, learning style and gender.
The analysis the analysis provides an overview of student learning behaviour and academic performance the observed difference between groups should be interpreted as a pattern the dataset is not as the proof that one factor directly causes changes in exam scores.

#limitations
-The dataset contains coded categorical variable show some categories cannot be interpreted by name without additional documentation.
-Correlation and group average show relationship of difference differences but do not prove cause and effect.
-The analysis is based on the variables available in the data set was not treated as an independent predictor because the strong relationship with ExamScore.
